```python
import numpy as np
import matplotlib.pyplot as plt
from uncertainties import ufloat, unumpy
```

```python
def FWHM(sigma):
    FWHM = 2*np.sqrt(2*np.log(2))*sigma
    # FWHM = 2.555*sigma
    return FWHM

def resolution(sigma, E):
    fwhm = FWHM(sigma)
    return fwhm/E, fwhm
```

```python
det1_mean = ufloat(511.7, 0.2)
det1_sigma = ufloat(110.84, 0.18)
res1, fwhm1 = resolution(det1_sigma, det1_mean)
print(res1, fwhm1)

det6_mean = ufloat(510.3,0.4)
det6_sigma = ufloat(23.39,0.46)
res6, fwhm6 =  resolution(det6_sigma, det6_mean)
print(res6, fwhm6)

print(np.diff(res6,res1))
print(fwhm6-fwhm1)
```
